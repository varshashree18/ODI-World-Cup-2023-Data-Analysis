```python
# Calculate total runs scored by each batsman
top_runs = (
    batting.groupby('Batsman_Name')['Runs']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_runs)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_runs.values,
    y=top_runs.index
)

plt.title('Top 10 Run Scorers - ODI World Cup 2023')
plt.xlabel('Total Runs')
plt.ylabel('Batsman')

plt.show()
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=top_runs.values,
    y=top_runs.index
)

plt.title('Top 10 Run Scorers - ICC ODI World Cup 2023')
plt.xlabel('Total Runs')
plt.ylabel('Batsman')

for i, value in enumerate(top_runs.values):
    plt.text(value + 5, i, str(value), va='center')

plt.show()
```

```python
# Calculate total wickets taken by each bowler
top_wickets = (
    bowling.groupby('Bowler_Name')['Wickets']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_wickets)
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=top_wickets.values,
    y=top_wickets.index
)

plt.title('Top 10 Wicket Takers - ICC ODI World Cup 2023')
plt.xlabel('Total Wickets')
plt.ylabel('Bowler')

# Add data labels
for i, value in enumerate(top_wickets.values):
    plt.text(value + 0.2, i, str(value), va='center')

plt.show()
```

```python
# Calculate total sixes hit by each batsman
top_sixes = (
    batting.groupby('Batsman_Name')['6s']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_sixes)
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=top_sixes.values,
    y=top_sixes.index
)

plt.title('Top 10 Players with Most Sixes - ICC ODI World Cup 2023')
plt.xlabel('Number of Sixes')
plt.ylabel('Batsman')

# Add data labels
for i, value in enumerate(top_sixes.values):
    plt.text(value + 0.2, i, str(value), va='center')

plt.show()
```

```python
# Calculate total runs and balls for each batsman
player_stats = batting.groupby('Batsman_Name').agg({
    'Runs': 'sum',
    'Balls': 'sum'
})

# Calculate overall strike rate
player_stats['Strike_Rate'] = (
    player_stats['Runs'] / player_stats['Balls'] * 100
).round(2)

# Filter players with at least 100 runs
highest_sr = (
    player_stats[player_stats['Runs'] >= 100]
    .sort_values('Strike_Rate', ascending=False)
    .head(10)
)

print(highest_sr)
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x=highest_sr['Strike_Rate'],
    y=highest_sr.index
)

plt.title('Top 10 Highest Strike Rate Players - ICC ODI World Cup 2023')
plt.xlabel('Strike Rate')
plt.ylabel('Batsman')

for i, value in enumerate(highest_sr['Strike_Rate']):
    plt.text(value + 1, i, str(value), va='center')

plt.show()
```
